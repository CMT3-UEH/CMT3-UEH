```python


files = glob.glob("data/*.parquet")

dfs = []
for f in files:
    df = pd.read_parquet(f)

    df["ticker"] = os.path.basename(f).replace(".parquet", "")

    dfs.append(df)

df_all = pd.concat(dfs, ignore_index=True)
print(df_all["ticker"].value_counts())
```
